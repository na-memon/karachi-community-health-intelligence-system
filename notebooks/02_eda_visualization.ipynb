## 3. Exploratory Data Analysis & Visualizations

```python
# import necessary libraries

import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Load the cleaned dataset
df = pd.read_excel(
    "../data/Karachi Community Health Surveillance Analytics System_cleaned.xlsx"
)
```

```python
# Ensure visuals directory exists
os.makedirs("../visuals", exist_ok=True)
```

```python
# Set global style for clean plots
sns.set_theme(style="whitegrid")
plt.rcParams.update({"figure.autolayout": True})

# --- FIGURE 1: Age Distribution ---
plt.figure(figsize=(8, 5))
sns.histplot(df["age"], bins=20, kde=True, color="teal")
plt.title(
    "Figure 1: Age Distribution of Survey Participants",
    fontsize=14,
    fontweight="bold",
)
plt.xlabel("Age (Years)", fontsize=12)
plt.ylabel("Frequency (Count)", fontsize=12)
plt.savefig("../visuals/fig1_age_distribution.png", dpi=300, bbox_inches="tight")
plt.show()

# --- FIGURE 2: BMI Classification ---
plt.figure(figsize=(8, 5))
bmi_order = ["underweight", "normal weight", "overweight", "obese"]
ax = sns.countplot(
    data=df,
    x="nutritional_scale",
    order=bmi_order,
    palette="Set2",
    hue="nutritional_scale",
    legend=False,
)
plt.title(
    "Figure 2: BMI Nutritional Status Classification",
    fontsize=14,
    fontweight="bold",
)
plt.xlabel("Nutritional Scale / BMI Category", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.xticks(
    ticks=[0, 1, 2, 3],
    labels=["Underweight", "Normal", "Overweight", "Obese"],
)
plt.savefig(
    "../visuals/fig2_bmi_classification.png", dpi=300, bbox_inches="tight"
)
plt.show()

# --- FIGURE 3: Gender Distribution ---
plt.figure(figsize=(6, 5))
sns.countplot(
    data=df, x="gender", palette="pastel", hue="gender", legend=False
)
plt.title(
    "Figure 3: Gender Breakdown of Participants", fontsize=14, fontweight="bold"
)
plt.xlabel("Gender", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.savefig(
    "../visuals/fig3_gender_distribution.png", dpi=300, bbox_inches="tight"
)
plt.show()

# --- FIGURE 4: Education Distribution ---
plt.figure(figsize=(8, 5))
sns.histplot(df["educational_years"], bins=15, discrete=True, color="royalblue")
plt.title(
    "Figure 4: Educational Years Distribution", fontsize=14, fontweight="bold"
)
plt.xlabel("Years of Education", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.savefig(
    "../visuals/fig4_education_distribution.png", dpi=300, bbox_inches="tight"
)
plt.show()

# --- FIGURE 5: Income Distribution ---
plt.figure(figsize=(8, 5))
sns.histplot(df["household_income"], bins=20, kde=True, color="darkorange")
plt.title(
    "Figure 5: Household Income Distribution", fontsize=14, fontweight="bold"
)
plt.xlabel("Household Income", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.savefig(
    "../visuals/fig5_income_distribution.png", dpi=300, bbox_inches="tight"
)
plt.show()

# --- FIGURE 6: Health Status ---
plt.figure(figsize=(8, 5))
# Check exact category strings present in your health_status column (e.g., excellent, very good, good, fair, poor)
health_order = ["excellent", "very good", "good", "fair", "poor"]
# Filter health_order to match existing values to avoid plotting errors if a category has 0 entries
existing_health = [h for h in health_order if h in df["health_status"].unique()]

sns.countplot(
    data=df,
    x="health_status",
    order=existing_health,
    palette="viridis",
    hue="health_status",
    legend=False,
)
plt.title(
    "Figure 6: Self-Reported Health Status Distribution",
    fontsize=14,
    fontweight="bold",
)
plt.xlabel("Health Status Category", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.savefig(
    "../visuals/fig6_health_status.png", dpi=300, bbox_inches="tight"
)
plt.show()
```
