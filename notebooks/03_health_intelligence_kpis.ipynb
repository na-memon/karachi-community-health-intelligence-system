## 3. Section 1: Population Overview KPIs

```python
import os
import pandas as pd
from IPython.display import HTML, display

# --- 1. LOAD THE PRE-CLEANED DATASET ---
df = pd.read_excel(
    "../data/Karachi Community Health Surveillance Analytics System_cleaned.xlsx"
)

# --- 2. CALCULATE POPULATION OVERVIEW KPIS ---
total_population = len(df)

# Use lowercase values as structured during data cleaning
male_count = (df["gender"] == "male").sum()
female_count = (df["gender"] == "female").sum()

# Calculate percentages
male_pct = round((male_count / total_population) * 100, 1)
female_pct = round((female_count / total_population) * 100, 1)

# Calculate averages
mean_age = round(df["age"].mean(), 1)
mean_bmi = round(df["bmi"].mean(), 1)

# Print metrics to console for log tracking
print("--- Population Overview KPIs Calculated Successfully ---")
print(f"Total Population: {total_population}")
print(
    f"Male: {male_count} ({male_pct}%) | Female: {female_count}"
    f" ({female_pct}%)"
)
print(f"Average Age: {mean_age} years")
print(f"Average BMI: {mean_bmi}")

# --- 3. RENDER VISUAL KPI CARDS IN JUPYTER ---
kpi_html = f"""
<div style="display: flex; gap: 15px; justify-content: space-between; font-family: Arial, sans-serif; margin: 20px 0;">
    
    <!-- Total Participants Card -->
    <div style="background: #f8fafc; border: 1px solid #e2e8f0; padding: 15px; border-radius: 8px; text-align: center; border-left: 5px solid #3182ce; flex: 1;">
        <h4 style="margin: 0; color: #4a5568; font-size: 13px;">Total Participants</h4>
        <p style="font-size: 22px; font-weight: bold; margin: 8px 0 0 0; color: #2b6cb0;">{total_population}</p>
    </div>

    <!-- Male % Card -->
    <div style="background: #f8fafc; border: 1px solid #e2e8f0; padding: 15px; border-radius: 8px; text-align: center; border-left: 5px solid #319795; flex: 1;">
        <h4 style="margin: 0; color: #4a5568; font-size: 13px;">Male %</h4>
        <p style="font-size: 22px; font-weight: bold; margin: 8px 0 0 0; color: #2c7a7b;">{male_pct}% <span style="font-size: 12px; color: #718096;">({male_count})</span></p>
    </div>

    <!-- Female % Card -->
    <div style="background: #f8fafc; border: 1px solid #e2e8f0; padding: 15px; border-radius: 8px; text-align: center; border-left: 5px solid #b794f4; flex: 1;">
        <h4 style="margin: 0; color: #4a5568; font-size: 13px;">Female %</h4>
        <p style="font-size: 22px; font-weight: bold; margin: 8px 0 0 0; color: #6b46c1;">{female_pct}% <span style="font-size: 12px; color: #718096;">({female_count})</span></p>
    </div>

    <!-- Average Age Card -->
    <div style="background: #f8fafc; border: 1px solid #e2e8f0; padding: 15px; border-radius: 8px; text-align: center; border-left: 5px solid #ed8936; flex: 1;">
        <h4 style="margin: 0; color: #4a5568; font-size: 13px;">Average Age</h4>
        <p style="font-size: 22px; font-weight: bold; margin: 8px 0 0 0; color: #c05621;">{mean_age} <span style="font-size: 12px; color: #718096;">yrs</span></p>
    </div>

    <!-- Average BMI Card -->
    <div style="background: #f8fafc; border: 1px solid #e2e8f0; padding: 15px; border-radius: 8px; text-align: center; border-left: 5px solid #48bb78; flex: 1;">
        <h4 style="margin: 0; color: #4a5568; font-size: 13px;">Average BMI</h4>
        <p style="font-size: 22px; font-weight: bold; margin: 8px 0 0 0; color: #276749;">{mean_bmi}</p>
    </div>

</div>
"""

display(HTML(kpi_html))
```

## 4. Section 2: Nutrition & BMI KPIs

```python
# --- SECTION 2: NUTRITION & BMI KPIS ---
total_pop = len(df)

# 1. Underweight Calculation (< 18.5)
underweight_count = (df['bmi'] < 18.5).sum()
underweight_rate = round((underweight_count / total_pop) * 100, 1)

# 2. Overweight Calculation (25 to < 30)
overweight_count = ((df['bmi'] >= 25) & (df['bmi'] < 30)).sum()
overweight_rate = round((overweight_count / total_pop) * 100, 1)

# 3. Obesity Calculation (>= 30)
obese_count = (df['bmi'] >= 30).sum()
obese_rate = round((obese_count / total_pop) * 100, 1)

# 4. Total Malnutrition Prevalence (Any deviation from normal weight: underweight + overweight + obese)
malnutrition_count = underweight_count + overweight_count + obese_count
malnutrition_rate = round((malnutrition_count / total_pop) * 100, 1)

# Print metrics to console for tracking
print('--- Nutrition & BMI KPIs Calculated Successfully ---')
print(
    f'Underweight Prevalence: {underweight_rate}% ({underweight_count} cases)'
)
print(f'Overweight Prevalence: {overweight_rate}% ({overweight_count} cases)')
print(f'Obesity Prevalence: {obese_rate}% ({obese_count} cases)')
print(
    f'Overall Malnutrition Prevalence: {malnutrition_rate}%'
    f' ({malnutrition_count} cases)'
)

# --- RENDER NUTRITION KPI CARDS IN JUPYTER ---
nutrition_kpi_html = f"""
<div style="display: flex; gap: 15px; justify-content: space-between; font-family: Arial, sans-serif; margin: 20px 0;">
    
    <!-- Malnutrition Prevalence Card -->
    <div style="background: #f8fafc; border: 1px solid #e2e8f0; padding: 15px; border-radius: 8px; text-align: center; border-left: 5px solid #e53e3e; flex: 1;">
        <h4 style="margin: 0; color: #4a5568; font-size: 13px;">Malnutrition Prevalence</h4>
        <p style="font-size: 22px; font-weight: bold; margin: 8px 0 0 0; color: #c53030;">{malnutrition_rate}% <span style="font-size: 12px; color: #718096;">({malnutrition_count})</span></p>
    </div>

    <!-- Underweight Card -->
    <div style="background: #f8fafc; border: 1px solid #e2e8f0; padding: 15px; border-radius: 8px; text-align: center; border-left: 5px solid #dd6b20; flex: 1;">
        <h4 style="margin: 0; color: #4a5568; font-size: 13px;">Underweight</h4>
        <p style="font-size: 22px; font-weight: bold; margin: 8px 0 0 0; color: #9c4221;">{underweight_rate}% <span style="font-size: 12px; color: #718096;">({underweight_count})</span></p>
    </div>

    <!-- Overweight Prevalence Card -->
    <div style="background: #f8fafc; border: 1px solid #e2e8f0; padding: 15px; border-radius: 8px; text-align: center; border-left: 5px solid #d69e2e; flex: 1;">
        <h4 style="margin: 0; color: #4a5568; font-size: 13px;">Overweight Prevalence</h4>
        <p style="font-size: 22px; font-weight: bold; margin: 8px 0 0 0; color: #b7791f;">{overweight_rate}% <span style="font-size: 12px; color: #718096;">({overweight_count})</span></p>
    </div>

    <!-- Obesity Prevalence Card -->
    <div style="background: #f8fafc; border: 1px solid #e2e8f0; padding: 15px; border-radius: 8px; text-align: center; border-left: 5px solid #9b2c2c; flex: 1;">
        <h4 style="margin: 0; color: #4a5568; font-size: 13px;">Obesity Prevalence</h4>
        <p style="font-size: 22px; font-weight: bold; margin: 8px 0 0 0; color: #742a2a;">{obese_rate}% <span style="font-size: 12px; color: #718096;">({obese_count})</span></p>
    </div>

</div>
"""

display(HTML(nutrition_kpi_html))
```

## 5. Section 3: Chronic Disease KPIs & Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns

# --- SECTION 3: CHRONIC DISEASE KPIS ---
total_pop = len(df)

# 1. Calculate counts and prevalence rates (%)
diabetes_count = (df['diabetes'] == 'yes').sum()
diabetes_rate = round((diabetes_count / total_pop) * 100, 1)

hypertension_count = (df['hypertension'] == 'yes').sum()
hypertension_rate = round((hypertension_count / total_pop) * 100, 1)

cancer_count = (df['cancer'] == 'yes').sum()
cancer_rate = round((cancer_count / total_pop) * 100, 1)

dyspepsia_count = (df['dyspepsia'] == 'yes').sum()
dyspepsia_rate = round((dyspepsia_count / total_pop) * 100, 1)

# Print metrics to console for tracking
print('--- Chronic Disease KPIs Calculated Successfully ---')
print(f'Hypertension Prevalence: {hypertension_rate}% ({hypertension_count} cases)')
print(f'Diabetes Prevalence: {diabetes_rate}% ({diabetes_count} cases)')
print(f'Dyspepsia Prevalence: {dyspepsia_rate}% ({dyspepsia_count} cases)')
print(f'Cancer Prevalence: {cancer_rate}% ({cancer_count} cases)')

# --- PLOT: DISEASE PREVALENCE (%) ---
diseases = ['Hypertension', 'Diabetes', 'Dyspepsia', 'Cancer']
rates = [hypertension_rate, diabetes_rate, dyspepsia_rate, cancer_rate]

plt.figure(figsize=(9, 5))
ax = sns.barplot(
    x=diseases, y=rates, palette='Reds_d', hue=diseases, legend=False
)

plt.title(
    'Figure 7: Chronic Disease Prevalence (%)', fontsize=14, fontweight='bold'
)
plt.xlabel('Health Conditions / Diseases', fontsize=12)
plt.ylabel('Prevalence Rate (%)', fontsize=12)

# Set dynamic Y-axis limit with padding for data labels
max_rate = max(rates) if max(rates) > 0 else 10
plt.ylim(0, max_rate * 1.25)

# Add exact percentage labels on top of each bar
for p in ax.patches:
  height = p.get_height()
  ax.annotate(
      f'{height}%',
      xy=(p.get_x() + p.get_width() / 2, height),
      xytext=(0, 4),  # 4 points vertical offset
      textcoords='offset points',
      ha='center',
      va='bottom',
      fontsize=10,
      fontweight='bold',
  )

# Save visualization to visuals folder
plt.savefig('../visuals/fig7_disease_prevalence.png', dpi=300, bbox_inches='tight')
plt.show()
```

## 6. Section 4: Lifestyle Risk KPIs

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- SECTION 4: LIFESTYLE RISK KPIS ---
total_pop = len(df)

print("--- Lifestyle Risk KPIs Calculated Successfully ---\n")

# 1. SMOKING STATUS BREAKDOWN
print("=== 1. Smoking Status Distribution ===")
smoking_counts = df['smoking_status'].value_counts()
smoking_pct = (df['smoking_status'].value_counts(normalize=True) * 100).round(1)

smoking_df = pd.DataFrame({'Count': smoking_counts, 'Percentage (%)': smoking_pct})
display(smoking_df)
print("\n")

# 2. SUBSTANCE ABUSE / ADDICTIVE SUBSTANCE FREQUENCY TABLE
print("=== 2. Addictive Substance Frequency Table ===")
if 'addictive_substance' in df.columns:
    sub_counts = df['addictive_substance'].value_counts()
    sub_pct = (df['addictive_substance'].value_counts(normalize=True) * 100).round(1)
    sub_df = pd.DataFrame({'Count': sub_counts, 'Percentage (%)': sub_pct})
    display(sub_df)
else:
    print("Column 'addictive_substance' not found.")
print("\n")

# 3. PHYSICAL ACTIVITY DISTRIBUTION
print("=== 3. Physical Activity Distribution ===")
if 'physical_activity' in df.columns:
    pa_counts = df['physical_activity'].value_counts()
    pa_pct = (df['physical_activity'].value_counts(normalize=True) * 100).round(1)
    pa_df = pd.DataFrame({'Count': pa_counts, 'Percentage (%)': pa_pct})
    display(pa_df)
    
    # Optional Plot for Physical Activity
    plt.figure(figsize=(7, 4))
    sns.countplot(data=df, x='physical_activity', palette='Set2', hue='physical_activity', legend=False)
    plt.title('Figure 8: Physical Activity Level Distribution', fontsize=13, fontweight='bold')
    plt.xlabel('Physical Activity Level', fontsize=11)
    plt.ylabel('Participant Count', fontsize=11)
    plt.savefig('../visuals/fig8_physical_activity.png', dpi=300, bbox_inches='tight')
    plt.show()
else:
    print("Column 'physical_activity' not found.")
```
