# Karachi Community Health Surveillance Analytics System (K-CHIS)

## Data Cleaning

```python
# import necessary libraries

import pandas as pd
import numpy as np
```

```python
# Load the dataset

df = pd.read_excel('../data/Karachi Community Health Surveillance Analytics System.xlsx')
```

```python
# clean the column names by converting to lowercase, replacing spaces with underscores, and removing parentheses

df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('(', '').str.replace(')', '')
df.columns
```

```python
for i in df.select_dtypes(include=['object', 'str']).columns:
    df[i] = df[i].astype(str).str.strip()  # Strip whitespace from string columns
```

```python
# Normalize NA values

na_values = ['NA', 'N/A', 'na', 'n/a', 'NaN', 'nan', 'null', 'NULL', '', ' ', None]

for i in df.columns:
    if df[i].dtype == 'object':
        df[i] = df[i].replace(to_replace= na_values, value= np.nan)  # Replace various NA representations with np.nan
```

```python
df['nutritional_scale'] = df['nutritional_scale'].str.replace('Normal weihgt', 'Normal weight', regex=False)
```

```python
# Automatically find object/string columns and convert them to lowercase
string_columns = df.select_dtypes(include=['object', 'str']).columns

for col in string_columns:
    df[col] = df[col].str.lower()
```

```python
# Check total missing values per column
print(df.isnull().sum())

# Check percentage of missing values per column
print((df.isnull().mean() * 100).round(2))
```

```python
# Handle 'dependents' (Numeric): Fill with median
if 'dependents' in df.columns:
    df['dependents'] = df['dependents'].fillna(df['dependents'].median())
```

```python
# Handle 'supplements' & 'wrokpreference' (Categorical): Fill with 'unknown'
categorical_fill_cols = ['supplements', 'wrokpreference']
for col in categorical_fill_cols:
    if col in df.columns:
        df[col] = df[col].fillna('unknown')
```

```python
# Handle 'addictive_substance' (High Missing): Fill with 'none' (assuming blank means no substance)
if 'addictive_substance' in df.columns:
    df['addictive_substance'] = df['addictive_substance'].fillna('none')
```

```python
# Verify remaining missing values
print("Remaining missing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Convert all text/object columns to lowercase and remove accidental extra spaces
string_cols = df.select_dtypes(include=['object', 'str']).columns
for col in string_cols:
    df[col] = df[col].str.lower().str.strip()

# Fix the specific "weihgt" typo in nutritional_scale if it exists
if 'nutritional_scale' in df.columns:
    df['nutritional_scale'] = df['nutritional_scale'].replace('normal weihgt', 'normal weight')

print("Text normalized successfully!")
```

```python
# Enforce Integer columns
int_cols = ['age', 'educational_years', 'household_income']
for col in int_cols:
    if col in df.columns:
        df[col] = df[col].astype(int)

# Enforce Float columns (including dependents now that the missing value is filled)
float_cols = ['height_m', 'weight_kg', 'bmi', 'dependents']
for col in float_cols:
    if col in df.columns:
        df[col] = df[col].astype(float)

# Enforce Categorical columns for all text-based variables
categorical_cols = [
    'gender', 'nutritional_scale', 'marital_status', 'ethinicity', 
    'family_status', 'living_status', 'economic_dependency', 'decision_making', 
    'health_status', 'hypertension', 'dyspepsia', 'cancer', 'diabetes', 
    'healthcare_visits', 'taking_medicines', 'take_care_during_illness', 
    'supplements', 'smoking_status', 'smoking_routine', 'types_of_beverages', 
    'substance_abuse', 'addictive_substance', 'physical_activity', 
    'servings6_11bread', 'serving2_4fruit', 'serving3_5vegetables', 
    'servings2_3dairy', 'serving2_3meat', 'fattype', 'householdactivity', 
    'wrokpreference', 'sleepingtime', 'sleepiingpills'
]

for col in categorical_cols:
    if col in df.columns:
        df[col] = df[col].astype('category')

# Print out final DataFrame info to verify everything looks right
print("\n--- Final Cleaned DataFrame Info ---")
print(df.info())
```

```python
# Check how many duplicate rows exist
print("Number of duplicate rows:", df.duplicated().sum())

# Drop duplicates and reset the index
df = df.drop_duplicates().reset_index(drop=True)

# Check the new shape of your DataFrame
print("Shape after dropping duplicates:", df.shape)
```

```python
# Check unique values for a few key columns
print("Gender unique values:", df['gender'].unique())
print("Hypertension unique values:", df['hypertension'].unique())
print("Nutritional scale unique values:", df['nutritional_scale'].unique())
```

```python
print("Final dataset shape:", df.shape)
```

```python
# Save fully cleaned data to Excel
df.to_excel('../data/Karachi Community Health Surveillance Analytics System_cleaned.xlsx', index=False)
```
